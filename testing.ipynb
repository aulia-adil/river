```python
from river import tree

update_model = tree.HoeffdingTreeClassifier(
        grace_period=200,
        leaf_prediction='nba'  # Use naive Bayes for simplicity
    )

from river.datasets import synth

dataset = synth.Agrawal(classification_function=0, seed=42)
test123 = 602
for i, (x, y) in enumerate(dataset.take(test123)):
    update_model.learn_one(x, y)

# take the data for 601 to 800
list_of_train_inputs = []
for i, (x, y) in enumerate(dataset.take(test123)):
    if i >= 600:
        list_of_train_inputs.append([x, y])
```

```python
# Check what nodes exist in the registry
print("Nodes in update_model registry:")
for node_id, node in update_model._node_registry.items():
    print(f"  Node {node_id}: {type(node).__name__}")
    if hasattr(node, 'stats'):
        print(f"    Stats: {node.stats}")
    if hasattr(node, 'splitter'):
        print(f"    Has splitter template: {node.splitter is not None}")
    if hasattr(node, 'splitters'):
        print(f"    Has splitters dict: {node.splitters is not None}")
        if node.splitters:
            print(f"    Features tracked: {len(node.splitters)}")
print()
```

```python
# Check the splitters for node 0
node_0 = update_model._node_registry.get(0)
print(f"Node 0 splitters dict: {node_0.splitters}")
print(f"\nNode 0 splitter template: {node_0.splitter}")
print(f"Node 0 splitter._att_dist_per_class: {node_0.splitter._att_dist_per_class}")
print()

# Check one of the actual splitters in the dict
if node_0.splitters:
    first_feature = list(node_0.splitters.keys())[0]
    print(f"Splitter for '{first_feature}': {node_0.splitters[first_feature]}")
    print(f"  _att_dist_per_class: {node_0.splitters[first_feature]._att_dist_per_class}")
```
